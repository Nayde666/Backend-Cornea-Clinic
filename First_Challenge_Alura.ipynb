## Importación de datos

```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

## 1. Análisis de facturación

```python
print(tienda['Producto'].count())
print(tienda2['Producto'].count())
print(tienda3['Producto'].count())
print(tienda4['Producto'].count())
```

### Ingreso total por tienda

```python
ingreso_tienda = tienda['Precio'].sum() # Para tienda 1
ingreso_tienda_dos = tienda2['Precio'].sum() # Para tienda 2
ingreso_tienda_tres = tienda3['Precio'].sum() # Para tienda 3
ingreso_tienda_cuatro = tienda4['Precio'].sum() # Para tienda 4

print(ingreso_tienda)
print(ingreso_tienda_dos)
print(ingreso_tienda_tres)
print(ingreso_tienda_cuatro)
```

## 2.Ventas por categoría

```python
def categorias(num_tienda):
  categorias = []
  for i in num_tienda['Categoría del Producto']:
    if i not in categorias:
      categorias.append(i)
  return categorias

categorias_t1 = categorias(tienda)
categorias_t2 = categorias(tienda2)
categorias_t3 = categorias(tienda3)
categorias_t4 = categorias(tienda4)

print(categorias_t1)
print(categorias_t2)
print(categorias_t3)
print(categorias_t4)
```

```python
def contar_categorias(num_tienda):
    conteo = {}
    for categoria in num_tienda['Categoría del Producto']:
        if categoria in conteo:
            conteo[categoria] += 1
        else:
            conteo[categoria] = 1
    return conteo

conteo_categoria_t1 = contar_categorias(tienda)
conteo_categoria_t2 = contar_categorias(tienda2)
conteo_categoria_t3 = contar_categorias(tienda3)
conteo_categoria_t4 = contar_categorias(tienda4)

print(conteo_categoria_t1)
print(conteo_categoria_t2)
print(conteo_categoria_t3)
print(conteo_categoria_t4)
```
